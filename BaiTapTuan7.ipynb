#### W7A1 Viết hàm tìm kiếm nhị phân trên một danh sách đã được sắp xếp tăng dần. Trả về chỉ số của phần tử hoặc -1 nếu không tìm thấy. mẫu hàm: def binary_search(arr, left, right, target):

```python
def binary_search(arr,left,right,target):
    while left<= right:
        mid = (left+right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid]<target:
            left = mid + 1
        else:
            right = mid -1
    return -1
arr = list(map(int,input().split()))
target = int(input())
print(binary_search(arr, 0, len(arr)-1, target))
```

#### W7A2 Cho danh sách arr và một số nguyên x, đếm xem x xuất hiện bao nhiêu lần trong danh sách. def count_occurrences(arr, x) -> int:

```python
def count_occurrences(arr,x):
    cnt = 0
    for j in arr:
        if j == x:
            cnt +=1
    return cnt
arr = list(map(int,input().split()))
x = int(input())
print(count_occurrences(arr,x))
```

#### W7A4 Viết hàm tìm phần tử có số lần xuất hiện nhiều nhất trong danh sách. Nếu có nhiều phần tử cùng số lần, trả về phần tử xuất hiện trước.


```python
a = list(map(int,input().split()))
d = {}
for x in a:
    if x in d:
        d[x]+=1
    else:
        d[x] = 1
b = max(d.values())
for j in d:
    if d[j] == b:
        print(j)
        break
    
```

#### W7A5 Cho danh sách các số nguyên và một số X, đếm tất cả các cặp (a, b) sao cho a + b = X


```python
lst = list(map(int,input().split()))
x = int(input())
cnt = 0
for i in range(len(lst)):
    for j in range(i+1,len(lst)):
        if (lst[i] + lst[j]) == x:
            cnt += 1
print(cnt)
```

#### W7A7 Cho một danh sách các số nguyên nums, hãy tìm độ dài của dãy con tăng nghiêm ngặt dài nhất (không nhất thiết phải liền kề).


```python
ds = eval(input())
do_dai = [1] * len(ds)
for i in range(len(ds)):
    for j in range(i):
        if ds[j] < ds[i]:
            do_dai[i] = max(do_dai[i], do_dai[j] + 1)
print(max(do_dai))
```

#### W7A8 Bạn được cung cấp: một danh sách các khoảng thời gian interval = [[l1, r1], [l2, r2], ...] và một danh sách các truy vấn queries = [q1, q2, ...] Với mỗi truy vấn q, hãy tìm độ dài nhỏ nhất của khoảng thời gian [li, ri] sao cho li ≤ q ≤ ri. Nếu không tồn tại khoảng thời gian nào như vậy, hãy trả về -1 cho truy vấn đó. solve in < O(m.n), hint: binary search in the queries list


```python
khoang = eval(input())
truy_van = eval(input())
ket_qua = []
for q in truy_van:
    nho_nhat = -1
    for l, r in khoang:
        if l <= q <= r:
            do_dai = r - l + 1
            if nho_nhat == -1 or do_dai < nho_nhat:
                nho_nhat = do_dai
    ket_qua.append(nho_nhat)
print(ket_qua)
```

#### W7A9 Cho một danh sách các số nums, với mỗi phần tử nums[i], hãy đếm xem có bao nhiêu phần tử bên phải nhỏ hơn nums[i]. Trả về danh sách các số đếm. Efficient solution: using heap or binary search tree

```python
nums = eval(input())
ket_qua = []
for i in range(len(nums)):
    dem = 0
    for j in range(i + 1, len(nums)):
        if nums[j] < nums[i]:
            dem += 1
    ket_qua.append(dem)
print(ket_qua)
```

#### W7A10 Cho một danh sách các số nguyên nums và một số nguyên m, hãy tìm tổng mảng con lớn nhất theo modulo m. Tức là, với bất kỳ mảng con nums[i:j] nào, hãy tính sum(nums[i:j]) % m và trả về kết quả lớn nhất có thể. Gợi ý: Duy trì tổng tiền tố (prefix sum) mod m và sử dụng tìm kiếm nhị phân để tìm tổng tiền tố nhỏ nhất chỉ lớn hơn tổng hiện tại một chút (để tối đa hóa hiệu số bao quanh).


```python
nums = eval(input())
m = int(input())
lon_nhat = 0
for i in range(len(nums)):
    tong = 0
    for j in range(i, len(nums)):
        tong += nums[j]
        if tong % m > lon_nhat:
            lon_nhat = tong % m
print(lon_nhat)
```
