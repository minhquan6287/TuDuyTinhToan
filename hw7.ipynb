#### W7A16 Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu mỗi phần tử x trong mảng xuất hiện đúng x lần. Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.


```python
ds = eval(input())
dem = {}
for x in ds:
    if x in dem:
        dem[x] += 1
    else:
        dem[x] = 1
xoa = 0
for x in dem:
    if dem[x] > x:
        xoa += dem[x] - x
    elif dem[x] < x:
        xoa += dem[x]
print(xoa)
```

#### W7A17 Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ phức tạp thời gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore. Nếu không có phần tử nào thỏa mãn thì không in gì


```python
a = eval(input())
x = None
d = 0
for i in a:
    if d == 0:
        x = i
        d = 1
    elif i == x:
        d += 1
    else:
        d -= 1
if a.count(x) > len(a) // 2:
    print(x)
```

#### W7A18Sắp xếp có điều kiện: Số chẵn đứng trước, số lẻ đứng sau, chẵn tăng dần, lẻ giảm dần; ; 


```python
a = eval(input())
c = []
l = []
for i in a:
    if i % 2 == 0:
        c.append(i)
    else:
        l.append(i)
c.sort()
l.sort()
print(c + l[::-1])
```

#### W7A19 Merge 2 mảng đã sắp xếp (có thể tăng dần, có thể giảm dần) mà không dùng bất cứ thuật toán sắp xếp nào


```python
a = eval(input())
b = eval(input())
kq = []
if a[0] < a[-1]:
    while a and b:
        if a[0] < b[0]:
            kq.append(a.pop(0))
        else:
            kq.append(b.pop(0))
else:
    while a and b:
        if a[0] > b[0]:
            kq.append(a.pop(0))
        else:
            kq.append(b.pop(0))
kq += a
kq += b
print(kq)
```

#### W7A20 Cho dãy N số nguyên dương, tính tổng các số có k chữ số. Dòng đầu ghi N số nguyên dương, dòng sau ghi số nguyên dương k


```python
a = list(map(int, input().split()))
k = int(input())
s = 0
for i in a:
    if len(str(i)) == k:
        s += i
print(s)
```
