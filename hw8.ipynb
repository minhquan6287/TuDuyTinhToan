#### Yêu cầu
Xây dựng lớp Shape với phương thức area().
Xây dựng lớp Rectangle(w, h) và Circle(r) kế thừa từ Shape, ghi đè area().
Lấy π = 3.14.
Input
Dòng 1: loại hình (RECTANGLE hoặc CIRCLE)
Dòng 2: tham số tương ứng.
Output
In ra diện tích với 2 chữ số thập phân

```python
class Rectangle:
    def __init__(self,w,h):
        self.w = w
        self.h = h
    def area(self):
        return self.w * self.h
class Circle:
    def __init__(self,r):
        self.r = r
    def area(self):
        return 3.14 * self.r * self.r
loai = input()
if loai == "Rectangle":
    w,h = map(float,input().split())
    hinh = Rectangle(w,h)
else:
    r = float(input())
    hinh = Circle(r)
print(hinh.area())

    
        
```
