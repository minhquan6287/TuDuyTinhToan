#### W8A2 [Date]
Xây dựng lớp Date biểu diễn ngày tháng năm với hai 
thức kiểm tra xem một ngày tháng nào đó có hợp lệ hay không
và phương thức tính ngày kế tiếp của ngày tháng năm nào đó
nếu nó là hợp lệ
Nhập vào một ngày tháng năm, (ngày tháng năm biểu diễn
dưới dạng dd/mm/yyyy). Nếu ngày tháng năm không tồn tại, in
ra dòng chữ INVALID. Nếu là hợp lệ in ra ngày kế tiếp của nó.
Đầu vào
Đầu vào từ bàn phim chứa dữ liệu ngày, tháng, năm dưới dạng
dd/mm/yyyy
.
Đầu ra
In ra màn hình ngày kế tiếp của ngày vừa nhập dưới dạng
dd/mm/yyyy
. Nếu ngày tháng không tồn tại, in ra INVALID 


```python
class Date:
    def __init__(self,d,m,y):
        self.d = d
        self.m = m
        self.y = y
    def checkValid(self):
        maxday = 0
        match self.m:
            case 1|3|5|7|8|10|12:
                maxday = 31
            case 4|6|9|11:
                maxday = 30
            case 2:
                if self.y%4 == 0 and self.y%100 != 0 or self.y %400 == 0:
                    maxday = 29
                else:
                    maxday = 28
        return 1<=self.d<= maxday and self.y >0
    def getnexday(self):
        nextday = Date(self.d +1,self.m,self.y)
        if nextday.checkValid():
            return nextday
        nextday.d = 1
        nextday.m += 1
        if nextday.checkValid():
            return nextday
        nextday.m = 1
        nextday.y += 1
        if nextday.checkValid():
            return nextday
d,m,y =map(int,input().split())
mydate1 = Date(d,m,y)
if not mydate1.checkValid():
    print("Invalid")
else:
    a = mydate1.getnexday()
    print(f"{a.d:02d}/{a.m:02d}/{a.y}")
```

#### W8A3 [Point]
Xây dựng lớp điểm với hai thuộc tính x, y cùng các phương
thức kiểm tra xem có phải gốc tọa dộ không, có nằm trên trục
tung không, có nằm trên trục hoành không, tính khoảng cách từ
điểm đó tới một điểm khác
Đầu vào:
Nhập vào tọa độ một điểm M bất kỳ
Đầu ra:
Thông báo M có phải gốc tọa độ không hoặc có nằm trên trục
tung hoặc trục hoành không, khoảng cách từ M tới gốc tọa độ
là bao nhiêu (lấy 2 chữ số phần thập phân) 


```python
class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    def isorigin(self):
        return self.x == 0 and self.y ==0
    def isox(self):
        return self.y == 0
    def isoy(self):
        return self.x ==0
    def distance(self):
        return ((self.x)**2+(self.y)**2)**0.5
x,y = map(float,input().split())
m = Point(x,y)
if m.isorigin():
    print("m la goc yoa do")
if m.isox():
    print("m nam tren ox")
if m.isoy():
    print("m nam tren oy")
print(f"{m.distance():.2f}")
```

#### W8A4 [Calculator]
Xây dựng lớp Calculator (máy tính bỏ túi) với các phương thức
để thực hiện các phép toán 1. add (cộng), 2. subtract(trừ), 3.
product (nhân) và 4. divide(chia), 5. power(hàm mũ), 6. mod
(lấy phần dư) và phương thức 7. set_numbers(cập nhật 2 số
mới) nếu người dùng muốn tính toán trên 2 số khác.
Khởi tạo lớp là hai thuộc tính gồm 2 số cần tính, Mỗi phương
thức thực hiện một phép tính và trả về kết quả tương ứng
Tạo menu và cho người dùng tính toán tới khi gõ "yes", "YES"
hoặc "Yes" cho câu hỏi "Bạn có muốn thoát?" mới dừng. 

```python
class calculator:
    def __init__(self,a,b):
        self.a = a
        self.b = b
    def add(self):
        return self.a + self.b
    def subtract(self):
        return self.a - self.b
    def product(self):
        return self.a * self.b
    def divive(self):
        return self.a / self.b
    def power(self):
        return self.a ** self.b
    def mod(self):
        if self.b ==0:
            return "Invalid"
        return self.a % self.b
a,op,b = input().split()
a = float(a)
b = float(b)
cal = calculator(a,b)
if op == "+":
    print(cal.add())
if op =="-":
    print(cal.subtract())
if op =="*":
    print(cal.product())
if op == "/":
    print(cal.divive())
if op == "^":
    print(cal.power())
if op =="%":
    print(cal.mod())
    
```

#### W8A5 [ShoppingCart]
Xây dựng lớp ShoppingCart (giỏ hàng) để quản lý các sản
phẩm của khách. Dùng danh sách để lưu các sản phẩm, mỗi
sản phẩm là một từ điển có name (tên) và price(giá).
Với các phương thức sau:
1. Khởi tạo với danh sách rỗng
2. Thêm vào giỏ hàng một sản phẩm mới
3. Xóa khỏi giỏ một sản phẩm nào đó (tìm theo tên để xóa)
4. Kiểm tra xem giỏ hàng có rỗng không
5. Tính tổng tiền giỏ hàng
6. Hiển thị toàn bộ sản phẩm trong giỏ
7. Xóa toàn bộ giỏ hàng
Tạo menu cho người dùng nhập lựa chọn truy vấn từ 2 tới 7.
Chương trình dừng khi người dùng nhập "yes" cho câu hỏi "Có
muốn thoát khỏi chương trình không?"

```python
class ShoppingCard:
    def __init__(self):
        self.cart = []
    def add_product(self,name,price):
        product = {"name":name,"price":price}
        self.cart.append(product)
    def remove_product(self,name):
        for x in self.cart:
            if x["name"] == name:
                self.cart.remove(x)
                return "Da xoa"
        return "Khong tim thay"
    def is_empty(self):
        return len(self.cart) == 0
    def price(self):
        s = 0
        for x in self.cart:
            s += x["price"]
        return s 
    def show_all(self):
        if self.is_empty():
            print("Gio hang rong")
        else:
            for x in self.cart:
                print(x["name"],"-",x["price"])
    def clear_cart(self):
        self.cart.clear()
cart = ShoppingCard()
while True:
    print("___")
    print("2.Them")
    print("3.Xoa")
    print("5.Tong tien")
    print("6.Hien thi")
    print("7.Xoa het")
    choice = input()
    if choice == "2":
        name = input()
        price = float(input())
        cart.add_product(name,price)
    elif choice == "3":
        name = input()
        cart.remove_product(name)
    elif choice == "5":
        print(cart.price())
    elif choice == "6":
        cart.show_all()
    elif choice == "7":
        cart.clear_cart()
    thoat = input("Thoat?")
    if thoat == "yes":
        break
```
